```python
df.columns = (
    df.columns 
    .str.strip()
    .str.lower()
    .str.replace(' ','_')
)
```

```python
df = df.drop(columns=[
    'transaction_hash',
    'order_id',
    'trade_id',
    'crossed'
])
```

```python
df['timestamp_ist'] = pd.to_datetime(
    df['timestamp_ist'],
    format = '%d-%m-%Y %H:%M'
)
```

```python
df['date'] =df['timestamp_ist'].dt.date
```

```python
df['side'] = df['side'].str.lower()
df['direction'] = df['direction'].str.lower()
```

```python
num_cols = [
    'execution_price',
    'size_tokens',
    'size_usd',
    'start_position',
    'closed_pnl',
    'fee'
]

df[num_cols] = df[num_cols].apply(pd.to_numeric, errors='coerce')
```

```python
df = df[df['closed_pnl'].notna()]
```

```python
df = df[df['closed_pnl'] != 0]
```

```python
df = df[
    (df['size_usd'] > 0) &
    (df['execution_price'] > 0)
]
```

```python
df['trade_volume'] = abs(df['size_usd'])
df['is_win'] = df['closed_pnl'] > 0
df['risk_exposure'] = df['size_usd'] * df['execution_price']
```

```python
df = df.drop(columns=['direction'])
```

```python
df['pnl_bucket'] = pd.cut(
    df['closed_pnl'],
    bins=[-1e9, 0, 500, 2000, 1e9],
    labels=['loss', 'small_win', 'medium_win', 'big_win']
)
```

```python
# Profitability flag
df["is_profitable"] = df["closed_pnl"] > 0

# ROI
df["roi"] = df["closed_pnl"] / df["size_usd"]

# Fee percentage
df["fee_pct"] = df["fee"] / df["size_usd"]
```

```python
df["date"] = df["timestamp_ist"].dt.date
df["hour"] = df["timestamp_ist"].dt.hour

daily_pnl = df.groupby("date")["closed_pnl"].sum()
hourly_pnl = df.groupby("hour")["closed_pnl"].mean()
```

```python
risk_pnl = df.groupby(pd.qcut(df["risk_exposure"], 5))["closed_pnl"].mean()
print(risk_pnl)
```

```python
import matplotlib.pyplot as plt

daily_pnl.plot(figsize=(12,5), title="Daily PnL Trend")
plt.show()
```

```python
risk_bucket = pd.qcut(df["risk_exposure"], q=5, duplicates="drop")

risk_pnl = (
    df
    .groupby(risk_bucket, observed=True)["closed_pnl"]
    .mean()
    .round(2)
)

print(risk_pnl)
```

```python
risk_summary = (
    df
    .groupby(risk_bucket, observed=True)
    .agg(
        avg_pnl=("closed_pnl", "mean"),
        trade_count=("closed_pnl", "count"),
        win_rate=("is_profitable", "mean")
    )
    .round(2)
)

print(risk_summary)
```
